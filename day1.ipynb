# Flow Control
Hoe gaat data door mijn programma heen?

Denk aan:
1. For loop (en list comprehension)
2. While loop
3. if / elif / else
4. break, continue, pass

```python
for index, placeholder in enumerate([1, 2, 3, 4, 5, 6, 7, 8, 9]):
    print(f'Value {placeholder} at position {index} in list')
```

```python
a = 42
y = 29

## if and elif
if a == 42 or y < 100:
    # do something
    pass

elif a < 42:
    # do something else
    pass

else:
    # do this other thing.
    pass
```

```python
## Nested loops
L = ["AGACTAGCT", "ADCTAD", "GCATCATGA"]

for dna in L:
    for nucleotide in dna:
        print(nucleotide)
```

```python
## Breaking and passing

fruits = ['mango', 'apple', 'pears', 'banana', 'kiwi']
for fruit in fruits:
    if fruit == 'apple':
        pass             # don't print apple. Using break here would only print mango.
    else:
        print(fruit)
```

```python
L = range(0,100)

for number in L:
    if number % 2 == 0:
        print(number)
    else:
        pass
```

```python
def fortytwoinator(arg1):
    """
    Beschrijving van de functie
    :param arg1:  float or list of floats
    :return:      float or list of floats where all values are 42.
    """
    if type(arg1) is float or type(arg1) is int:
        return 42
    elif type(arg1) is list:
        for index in range(len(arg1)):
            arg1[index] = 42
        return arg1
    else:
        print("Error: supplied argument is neither a float nor a list.. but have your result anyway.")
        return 42

A = fortytwoinator(13)
B = fortytwoinator([1, 2, 3, 4])

print(A)
print(B)

```

```python
## Args and Kwargs
def myfunction(*args, **kwargs):
    for item in args:  # Function unpacks the given arguments as a tuple
        print(item)

    for key in kwargs: # Function unpacks the given keyword arguments as a dictionary
        print(key, kwargs[key])

    return 42

myfunction(1, 2, 3, x=5, y=6, z=7)
```

# Dunder variable __name__
De dunder variable `__name__` checkt informatie over een script.
Als het script zelf gerund wordt, bevat deze variabele `__main__`. Als het script in plaats daarvan wordt aangeroepen in een ander script, dan bevat het dit niet. Daarom zie je vaak een dergelijke constructie in scripts die functionaliteit uit andere scripts gebruiken:

`if __name__ == "__main__": `

`    main()`




```python
def main():
    pass

if __name__ == "__main__":
    main()
```
