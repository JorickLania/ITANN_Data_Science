### Question 1

Which country has the highest estimate of TB?

This function should return a the string India.

```python
def answer_one(df):
    highest_TB_country = df.loc[df['hi'].idxmax()].country
    #df.sort_values('best', ascending=False, inplace=False)

    return highest_TB_country

print(f'Country with highest estimate of TB: {answer_one(df)}')
```

### Question 2

Which country has the smallest difference between high and low estimates?

```python
def answer_two(df):
    diff_high_low = df['hi'] - df['lo'] #panda series of the differences
    smallest_diff_country = df.loc[diff_high_low.idxmin()].country
    
    return smallest_diff_country #write your code here, should return American Samoa

answer_two(df)
```

### Question 3

How many countries have 65plus records?

```python
def answer_three(df):
    countries_with_65plus = df.loc[df['age']=='65plus'].country
    unique_countries = countries_with_65plus.unique()
    return unique_countries.size

answer_three(df)
```

### Question 4

Select the females and exclude the 'age = all' records. What are the top 5 countries that have the two most highest 'best estimates'? 


```python
#write your code here, should return this
"""
India          1084000
Indonesia       393000
China           296000
Pakistan        252000
Philippines     173000
"""

df_subset = df.loc[(df['sex']=='f') & (df['age'] != 'all')] # select the females and exclude the age=all records
df_f_bycountry = df_subset.groupby('country')               # grouped by country
two_bests = df_f_bycountry['best'].nlargest(2)
countries_twobest_summed = two_bests.groupby('country').sum()

highest_countries = countries_twobest_summed.nlargest(5)

print(highest_countries)
```
