```python
import pandas as pd
import numpy as np
```

```python
df_raw = pd.read_table('data/JamesBondGlucoseLevels25.txt',
                       )
df_raw.head()
```

```python
## Converting the time column to datetime format
df_raw.Time = pd.to_datetime(df_raw.Time)
df_raw = df_raw.set_index('Time') # Set time as the index; useful for plotting
```

```python
## Handling the NaNs and space values
df_raw.replace('  ', np.nan, inplace=True)

## Now convert string column to numeric
df_raw['glucose_scan'] = pd.to_numeric(df_raw['Scan.glucose.mg.dL.']) # create new column and put the converted vals into it
```

```python
df_raw.head()
```

```python
## Put our work in a function

def reformat():
    df = pd.read_table('data/JamesBondGlucoseLevels25.txt')
    df.Time = pd.to_datetime(df.Time)
    df = df.set_index('Time')
    df.replace('  ', np.nan, inplace=True)
    df['glucose_scan'] = pd.to_numeric(df['Scan.glucose.mg.dL.'])
    df.drop('Scan.glucose.mg.dL.', axis=1, inplace=True)

    return df
```

```python
df = reformat()
df.info()
```

# Datasets Combineren
- pd.concat: rijen / kolommen aan elkaar plakken
- pd.merge: meer SQL-achtige syntax, meer opties

```python
## Several methods
# pd.concat
# pd.merge
# pd.DataGrame.join

df_01 = pd.DataFrame(np.random.randn(3,5))
df_02 = pd.DataFrame(np.random.randn(3,5))
```

```python
df_03 = pd.concat([df_01, df_02])
df_03
```

Houd er rekening mee dat .concat hier de indexes onder elkaar plakt; indices hoeven dus niet meer uniek te zijn.
We kunnen ook op kolombasis de tabellen concateneren.

```python
df_04 = np.concat([df_01, df_02], axis=1)
df_04
```

### pd.merge

```python
left = pd.DataFrame({'key': ['A', 'B', 'C'], 'left_value' : [1, 2, 3]})
right = pd.DataFrame({'key': ['A', 'B', 'X'], 'right_value' : [3, 4, 5]})
```

```python
pd.merge(left, right, how='inner', left_on='key', right_on='key')
```

```python
pd.merge(left, right, how='outer', left_on='key', right_on='key')
```

### Pivot and Melt - dataframes reshapen

```python
data = {
    "value" : range(12),
    "variable" : ['A'] * 3 + ['B'] * 3 + ['C'] * 3 + ['D'] * 3,
    "date" : pd.to_datetime(['2020-01-01', '2020-01-02', '2020-01-03'] * 4),
}

df = pd.DataFrame(data)
```

```python
df
```

Nu willen we 'A', 'B', 'C', 'D' als kolom headers hebben, en de getallen als de waarden in de rows. Dit kun je doen met Pivot.

(long format to wide format)

```python
pivoted = df.pivot(index='date', columns='variable', values='value')
pivoted
```

```python
## Wat als ik meer kolommen heb?
df['value2'] = df['value'] * 2
df['value3'] = df['value'] * 3
df
```

```python
hierarchical_table = df.pivot(index='date', columns='variable')
hierarchical_table
```

```python
## Access it again:
hierarchical_table['value2']
```

Melt is het omgekeerde: gaat van een wide-format naar een long-format.
