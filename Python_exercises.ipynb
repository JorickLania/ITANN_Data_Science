```python
def partial_volume(radius):
    """
    Calculates partial volume given a radius.
    :param radius:      float or int
    :return:            partial volume (float)
    """
    return (2/3)*(math.pi*(radius**3))

a = (4.6+1.2)/2.09
b = 3*(4**(7/12))
pv = partial_volume(6)
c = 5*((4+2)/(9*math.sqrt(7)))

print(f'Answers: \n{a}\n{b}\n{pv}\n{c}')
```
