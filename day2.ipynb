### List Comprehension


```python
A = [x**2 for x in range(1,10)]
print(A)

B = []
for x in range(1,10):
    B.append(x**2)

print(B)
```

Voor kleine lists is de tweede methode hierboven sneller; er zit wat overhead in het maken van een comprehension.
Maar voor langere lijsten, of met duurdere operations, is list comprehension sneller, ookal is de tweede ook leesbaarder.

```python
data = ['Hi!', 42, 120, 6, True]
[x**2 if type(x) == 'int' else x*2 for x in data]
```

### Try and Except
Errors opvangen zodat je code met voorspelbare errors kan omgaan.

```python
L = [1, 2, 3, 4, 5]
try:
    print(L[5])  # Will raise an error, since index 5 is out of bounds for a list of len 5
except:
    print('Nuuuuu! An error!') # Catches ALL errors - not just the Index Out of Bounds!


## Nicer:

try:
    print(L[5])
except IndexError:
    lastval = L[-1]
    print(f"That index is out of bounds, but here is the closest preceding value: {lastval}")
except TypeError:
    print("Encountered a TypeError.")
```

### Object-Oriented Programming


```python
## Classes - what does x have, what can x do?

class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        self.name_upper()

    def name_upper(self):                  # A method is a function defined within a class, which can be applied to instances
        self.allcaps = self.name.upper()
        return self.allcaps                # of that class. So we can call studentinstance.allcaps

    def __str__(self):
        return f'Instance of Student class\nAttributes: {self.__dict__}'

Jorick = Student('Jorick', 25)
print(Jorick.name, Jorick.age)

## Note that each class inherits the template of an Object class; hence a bunch of dunder methods are already built in.

## You can add things to a class at any time; but adding it to an instance won't add it to other instances.

Marloes = Student('Jorick', 25)

Jorick.cijfers = [8, 7.5, 9.5, 8.5, 10,0]


print(Jorick.cijfers)

try:
    print(Marloes.cijfers)
except AttributeError:
    print("Instance of Student Object 'Marloes' has no attribute 'cijfers'.")

print(f'Trying out our method: {Jorick.allcaps}')
```

### Pandas and NumPy
- Turning data into information: "Munging"
- Turning information into knowledge: "Analysis"

- Tidy data: ground level; in a form amenable to data science
- FAIR data: at a high level' in form and storage allowing it to carry on beyond its initial scope, project and group

- R programming language set standard for Tidy Data; book _R for Data Science_ is useful to read even for Python developers.
- Three rules: every variable has a column, every observation has a row, every value in its own cell.

- Codebooks: describes the data; its columns, column descriptions, data types, etcetera - everything someone else needs to understand this dataset and take it up for their own use.

### Fair data:
also known as data stewardship. Meets principles of Findability, accessibility, interoperability, and reusability. Answers the question: _What to do with data once you've tidied, analyzed and published it?_ --> You make it available for others!
